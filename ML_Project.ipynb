```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

# Importing The Dataset

```python
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]

columns = ['CRIM','ZN','INDUS','CHAS','NOX','RM','AGE','DIS','RAD',
           'TAX','PTRATIO','B','LSTAT']
df = pd.DataFrame(data, columns=columns)
df['MEDV'] = target

print(df.head())
```

```python
df=df.rename(columns={'MEDV':'Price'})
```

## Independent & Dependent Features

```python
X = df.iloc[:,df.columns!='Price']
y = df.iloc[:,df.columns=='Price']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train ,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_test.shape
```

```python
 # Standarize the dataset
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

## Model Training

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
## The coefficients and intercept
print(f"Coeficients >> {model.coef_ }")
print(f"intercept >> {model.intercept_ }")
```

```python
model.get_params()
```

```python
## prediction of the model
y_prid = model.predict(X_test)
```

```python
y_prid
```

```python
plt.scatter(y_test, y_prid)
```

```python
error = y_test - y_prid
```

```python
error
```

```python
## plt the error
sns.displot(error,kind="kde")
```

```python
plt.scatter(y_prid,error)
```

## Model Metrics

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error,r2_score

print(f"MSE >> {mean_squared_error(y_test,y_prid)}")
print(f"RMSE >> {np.sqrt(mean_squared_error(y_test,y_prid))}")
print(f"MAE >> {mean_absolute_error(y_test,y_prid)}")
```

```python
score = r2_score(y_test,y_prid)
print(f"R2 Score >> {score}")
```
